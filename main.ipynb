```python
import os
import re

def extract_subject_ids(folder_path):
    ids = set()
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            match = re.match(r"(ACLR_\d{3})", file) or re.match(r"(Halo_\d{3})", file) or re.match(r"(PT_\d{3})", file)
            if match:
                ids.add(match.group(1))
    return ids
```

```python

import os
import pandas as pd
import re

# Define paths again

csv_path = "/home/zip22003/AMI-AI/dat files_emg data/Extracted_AMI_values.csv"
extract_dir = "/home/zip22003/AMI-AI/dat files_emg data"


# List .dat files
extracted_files = []
for root, dirs, files in os.walk(extract_dir):
    for file in files:
        if file.endswith(".dat"):
            extracted_files.append(os.path.join(root, file))

# Build the initial dataset
data_entries = []
for path in extracted_files:
    filename = os.path.basename(path)
    subject_match = re.search(r"(ACLR_\d+)", filename) or re.search(r"(Halo_\d+)", filename) or re.search(r"(PT_\d+)", filename)
    subject = subject_match.group(1) if subject_match else None
    label = None
    if "Injured" in path:
        label = "ACLR_AMI_Y_1_N_0"
    elif "Contra_Limb" in path:
        label = "Contra_AMI_Y_1_N_0"
    if subject and label:
        data_entries.append({
            "file_path": path,
            "subject": subject,
            "label": label
        })
dataset_df = pd.DataFrame(data_entries)


```

```python
dataset_df
```

```python
ami_df = pd.read_csv(csv_path)
ami_df.head()
```

```python
# Merge dataset with AMI values
merged_df = dataset_df.merge(ami_df, left_on="subject", right_on="Subject", how="left")

# Replace the "label" string with actual numeric values from AMI CSV
def get_numeric_label(row):
    if row["label"] == "ACLR_AMI_Y_1_N_0":
        return row["ACLR_AMI_Y_1_N_0"]
    elif row["label"] == "Contra_AMI_Y_1_N_0":
        return row["Contra_AMI_Y_1_N_0"]
    else:
        return None

merged_df["numeric_label"] = merged_df.apply(get_numeric_label, axis=1)
merged_df = merged_df[["file_path", "subject", "label", "numeric_label"]]

```

```python
filtered_df = merged_df.dropna(subset=["numeric_label"])
```

```python
import os
import csv

vlo_folder = "/home/zip22003/AMI-AI/dat files_emg data/Healthy/Injured_Matched_Limb/VLO"
vmo_folder = "/home/zip22003/AMI-AI/dat files_emg data/Healthy/Injured_Matched_Limb/VMO"

vlo_files = {f for f in os.listdir(vlo_folder) if f.endswith(".dat")}
vmo_files = {f for f in os.listdir(vmo_folder) if f.endswith(".dat")}

pairs = []
unpaired_vlo = []
unpaired_vmo = []

# 遍历 VLO → 找对应的 VMO
for fname in vlo_files:
    vmo_fname = fname.replace("VLO", "VMO")
    if vmo_fname in vmo_files:
        pairs.append((os.path.join(vlo_folder, fname),
                      os.path.join(vmo_folder, vmo_fname)))
    else:
        unpaired_vlo.append(fname)

# 遍历 VMO → 找对应的 VLO
for fname in vmo_files:
    vlo_fname = fname.replace("VMO", "VLO")
    if vlo_fname not in vlo_files:
        unpaired_vmo.append(fname)
```

```python
output_csv = "/home/zip22003/AMI-AI/vlo_vmo_pairs4.csv"
with open(output_csv, "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["VLO_path", "VMO_path"])

    for vlo, vmo in pairs:
        writer.writerow([vlo, vmo])
```

```python
output_csv = "/home/zip22003/AMI-AI/vlo_vmo_pairs5.csv"
with open(output_csv, "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["VLO_path", "VMO_path"])

    for vlo, vmo in zip(unpaired_vlo, unpaired_vmo):
        writer.writerow([os.path.join(vlo_folder, vlo), os.path.join(vmo_folder, vmo)])
```

```python
import re
import pandas as pd

ami_df = pd.read_csv("/home/zip22003/AMI-AI/Extracted_AMI_values.csv")
pairs_df = pd.read_csv("/home/zip22003/AMI-AI/vlo_vmo_pairs.csv")





# 从 ami_df 的 Subject 列获取所有 subject 值
valid_subjects = {s.upper() for s in ami_df["Subject"].dropna().unique()}

def extract_subject(path):
    if isinstance(path, str):
        path_upper = path.upper()  # 文件路径也转成大写
        for subj in valid_subjects:
            if subj in path_upper:
                return subj  # 返回统一后的大写 subject
    return None

pairs_df["subject"] = pairs_df["VLO_path"].apply(extract_subject)
```

```python
pairs_df
```

```python
pairs_df_cleaned = pairs_df.dropna(subset=["subject"])
```

```python
merged = pairs_df_cleaned.merge(ami_df, left_on="subject", right_on="Subject", how="left")

# Assign y_label according to rules
def assign_y(row):
    if isinstance(row["VLO_path"], str):
        if "injured" in row["VLO_path"]:  # injured
            return row.get("ACLR_AMI_Y_1_N_0", None)
        if "Contra_Limb" in row["VLO_path"]:  # Contra_Limb
            return row.get("Contra_AMI_Y_1_N_0", None)
    return None

merged["y_label"] = merged.apply(assign_y, axis=1)
```
